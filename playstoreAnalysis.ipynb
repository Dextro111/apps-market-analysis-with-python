## Google Play Store Datasets Analysis with Python

### Making Use of User Defined Python Functions, to explore, clean, and Interpret the Dataset

```python
# Opening, Accesing the file and Printing the Header Columns

from csv import reader
### The google Play Dataset ###
opened_file = open('googleplaystore.csv')
read_file = reader(opened_file)
android = list(read_file)
android_header = android[0]
android = android[1:] 
```

```python
# function to display a subset of the Dataset
def explore_data(dataset, start, end, rows_and_columns=False):
  dataset_slice = dataset[start:end]
  for row in dataset_slice:
    print(row)
    print('\n')
    
  if rows_and_columns:
    print('Number of rows', len(dataset))
    print('Number of Column', len(dataset[0]))
    
explore_data(android, 0,6, rows_and_columns=True)
```

```python
# Loop to check for Duplicate Apps
duplicate_apps = []
unique_apps = []

for app in android:
  name = app[0]
  if name in unique_apps:
    duplicate_apps.append(name)
  else:
    unique_apps.append(name)
    
    
# Loop to view some duplicate apps
for name in duplicate_apps[5:8]:
  for row in android:
    if name == row[0]:
      print(row)
      
print(f"Number of Duplicates are {len(duplicate_apps)}" )
```

```python
# The Deduplication Process. We keep the Row with the max Reviews and discard the rest
reviews_max = {}

for app in android:
  name = app[0]
  n_reviews = float(app[3]) # Zero Indexing
  
  if name in reviews_max and reviews_max[name] < n_reviews:
    reviews_max[name] = n_reviews
  elif name not in reviews_max:
    reviews_max[name] = n_reviews
    
print(len(reviews_max))
```

```python
# Viewing the Cleaned Android dataset

android_clean = []
already_added = []

for app in android:
  name = app[0]
  n_reviews = float(app[3])
  
  if (n_reviews == reviews_max[name]) and name not in already_added:
    android_clean.append(app)
    already_added.append(name)
    
print(android_clean[4:8])
```

```python
# Checking Non English Characters, and removing them 

def english_chars(name):
  bad_char_count = 0
  for char in name:
    if ord(char) > 127:
      bad_char_count += 1
    
    if bad_char_count > 3:
      return False
      
  return True

test_list = ['instagram','Docs To Go™ Free Office Suite', 'Instachat😜', '爱奇艺PPS -《欢乐颂2》电视剧热播'] 

for name in test_list:
  print(english_chars(name))
```

```python
# Seperating the Apps name Into English and Non English
english_android = []
non_english_android = []

for app in android_clean:
  name = app[0]
  
  if english_chars(name):
    english_android.append(app)
  else:
    non_english_android.append(app)
    
for app in non_english_android:
  print(app[0]) 
```

```python
# Filtering For Free English Androids Apps

free_eng_androids = []

for app in english_android:
  price = app[7]
  
  if (price == '0') or (price == '0.0'):
    free_eng_androids.append(app)
    
print(len(free_eng_androids))
```

```python
# Creating A frequency Table
def frequency_table(dataset, index): # Index refers  to the column 
  table = {}
  total = 0
  
  for row in dataset:
    total += 1
    value = row[index]
    if value in table:
      table[value] += 1
    else:
      table[value] = 1
      
  table_percentage = {}
  for key in table:
    percentage = (table[key] / total) * 100
    table_percentage[key] = percentage
    
  return table_percentage 

frequency_table(free_eng_androids,1)
```

```python
def display_table(dataset, index):
   table = frequency_table(dataset, index)
   table_display = []
   for key in table:
      key_val_as_tuple = (table[key],key)
      table_display.append(key_val_as_tuple)
     
   table_sorted = sorted(table_display, reverse=True)
   for entry in table_sorted:
      print(entry[1], ':', entry[0])
      
display_table(free_eng_androids, 1) # By category
```

```python
display_table(free_eng_androids, 5) # By installs
```

```python
categories_android = frequency_table(free_eng_androids, 1)
best_category = ''
best_avg_install_rate = 0

for category in categories_android:
    total = 0
    len_category = 0
    for app in free_eng_androids:
        category_app = app[1]
        if category_app == category:            
            n_installs = app[5]
            n_installs = n_installs.replace(',', '')
            n_installs = n_installs.replace('+', '')
            total += float(n_installs)
            len_category += 1
    
    avg_n_installs = total / len_category
    
    if avg_n_installs > best_avg_install_rate:
        best_category = category
        best_avg_install_rate = avg_n_installs
    
    print(category, ':', avg_n_installs)
    
print('\n\n')
print(f"Best Category: {best_category} \nAverage Installs: {best_avg_install_rate}")
```
