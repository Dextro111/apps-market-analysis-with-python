## Google Play Store Datasets Analysis with Python

### Making Use of User Defined Python Functions, to explore, clean, and Interpret the Dataset

```python
# Opening, Accesing the file and Printing the Header Columns

from csv import reader
### The google Play Dataset ###
opened_file = open('googleplaystore.csv')
read_file = reader(opened_file)
android = list(read_file)
android_header = android[0]
android = android[1:] 
```

```python
android_header
```

```python
# function to display a subset of the Dataset
def explore_data(dataset, start, end, rows_and_columns=False):
  dataset_slice = dataset[start:end]
  for row in dataset_slice:
    print(row)
    print('\n')
    
  if rows_and_columns:
    print('Number of rows', len(dataset))
    print('Number of Column', len(dataset[0]))
    
explore_data(android, 0,6, rows_and_columns=True)
```

```python
explore_data(android, 10470, 10475, rows_and_columns=True)
```

```python
# Loop to check for Duplicate Apps
duplicate_apps = []
unique_apps = []

for app in android:
  name = app[0]
  if name in unique_apps:
    duplicate_apps.append(name)
  else:
    unique_apps.append(name)
    
    
# Loop to view some duplicate apps
for name in duplicate_apps[5:8]:
  for row in android:
    if name == row[0]:
      print(row)
      
print(f"Number of Duplicates are {len(duplicate_apps)}" )
```

```python
# The Deduplication Process. We keep the Row with the max Reviews and discard the rest
reviews_max = {}

for app in android:
  name = app[0]
  n_reviews = float(app[3]) # Zero Indexing
  
  if name in reviews_max and reviews_max[name] < n_reviews:
    reviews_max[name] = n_reviews
  elif name not in reviews_max:
    reviews_max[name] = n_reviews
    
print(len(reviews_max))
```
